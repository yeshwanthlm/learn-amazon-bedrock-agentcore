### Required
#### AWS Account with appropriate IAM permissions for:
* Amazon Bedrock
* Amazon Cognito (for OAuth)
* AWS Lambda (for runtime deployment)
* CloudWatch (for monitoring)
* Python 3.10+ installed on your system
* uv package manager - Installation guide
* Basic knowledge of Python and AWS concepts

#### 1. Create Free AWS Account: https://aws.amazon.com/free/

#### 2. Create AWS Builder ID Account: https://aws.amazon.com/builder/

#### 3. Download Kiro (Free AI Powered Code Editor): https://kiro.dev/downloads/

#### 4. Clone the GitHub Repo:
```git clone https://github.com/yeshwanthlm/learn-amazon-bedrock-agentcore.git```

#### 5. Install Python and UV: https://docs.astral.sh/uv/getting-started/installation/#installation-methods or run the installation script: setup.sh

#### 6. Create AWS IAM User: ```learn-agentcore``` and add PowerUserAccess

#### 7. Configure AWS CLI on your terminal: https://youtu.be/JU8PCPIVQcE?si=cMEwluWSqSadLwd1
